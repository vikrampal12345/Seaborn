# 1. Relational Plot (figure level)



## a. scatterplot  (axes level)
## b. lineplot     (axes level)
## c. relplot (general form for using both scatterplot and lineplot)  (figure level)

relplot() → Figure-level plot

Purpose:

✓ Relationship between variables

✓ Correlation analysis

✓ Trend analysis

Kinds:
1. scatter → Numeric vs Numeric
2. line → Trend / Time-series data

Extra Features:

hue   → Color groups

style → Marker styles

size  → Marker size

Example:

sns.relplot(data=tips,
            x='total_bill',
            y='tip',
            hue='sex',
            style='time',
            size='size')
