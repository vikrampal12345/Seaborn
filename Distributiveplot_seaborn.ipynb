## Distribution Plot (Distplot) – Short Notes

### Definition

A distribution plot visualizes how the values of a numerical variable are distributed across a dataset.

### Purpose

* Understand the shape of the data
* Identify the center and spread
* Detect skewness and outliers
* Check whether the data follows a normal distribution

### Types of Distribution Plots

1. **Histogram**

   * Displays the frequency of values using bins (bars).
   * Best for understanding data spread and frequency.

2. **Kernel Density Estimate (KDE) Plot**

   * Shows a smooth probability density curve.
   * Helps visualize the overall shape of the distribution.

3. **Histogram + KDE Plot**

   * Combines frequency bars with a smooth density curve.
   * Most commonly used distribution plot.

4. **Rug Plot**

   * Displays individual data points as small lines on an axis.
   * Useful for observing data concentration.

5. **Empirical Cumulative Distribution Function (ECDF) Plot**

   * Shows the cumulative proportion of observations.
   * Useful for comparing distributions.

### Common Distribution Shapes

* **Normal Distribution:** Symmetric bell-shaped curve
* **Right-Skewed (Positive Skew):** Long tail on the right
* **Left-Skewed (Negative Skew):** Long tail on the left
* **Uniform Distribution:** Values are evenly distributed
* **Bimodal Distribution:** Two distinct peaks

### Seaborn Functions

```python
sns.histplot(data=df, x='column_name', kde=True)
sns.kdeplot(data=df, x='column_name')
sns.rugplot(data=df, x='column_name')
sns.ecdfplot(data=df, x='column_name')
```

> `sns.distplot()` is deprecated in recent versions of Seaborn.

### Important Parameters

* `bins` → Number of intervals in the histogram
* `kde=True` → Adds a KDE curve
* `color` → Changes plot color
* `stat='density'` → Displays density instead of count

### Example

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=df, x='Age', kde=True)
plt.show()
```

### Use Cases

* Exploratory Data Analysis (EDA)
* Feature engineering
* Outlier detection
* Checking normality assumptions
* Choosing appropriate data transformations
