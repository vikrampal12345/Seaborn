# Seaborn: Complete Notes

## What is Seaborn?

Seaborn is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for creating attractive, informative, and statistical graphics with less code. Seaborn works seamlessly with Pandas DataFrames and is widely used in Exploratory Data Analysis (EDA) and Machine Learning projects.

### Advantages of Seaborn

* Attractive default styles and themes
* Less code compared to Matplotlib
* Built-in statistical visualizations
* Easy integration with Pandas
* Helpful for data analysis and machine learning

# Main Classification of Seaborn Plots

Seaborn plots can be divided into five major categories:

### Purpose

Used to visualize relationships between two or more variables.

### Functions

* scatterplot()
* lineplot()
* relplot()

### Use Cases

* Finding trends
* Detecting correlations
* Comparing variables

# Seaborn relplot() Notes

## What is relplot()?

`relplot()` is a Figure-Level Function in Seaborn used to visualize relationships between variables. It can create Scatter Plots and Line Plots and allows multiple dimensions of data to be displayed in a single graph.

### Syntax

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    kind='scatter',
    hue='sex',
    style='time',
    size='size'
)
```

---

# Important Parameters

## 1. kind

### Purpose

Determines the type of relational plot.

### Common Values

* `'scatter'` → Scatter Plot
* `'line'` → Line Plot

### Example

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

### Use Case

Used to choose the visualization type according to the data.

---

## 2. hue

### Purpose

Changes the color of points based on a categorical variable.

### Example

```python
hue='sex'
```

### Output

* Male → One color
* Female → Another color

### Use Case

Helps compare different categories using colors.

---

## 3. style

### Purpose

Changes the marker shape based on categories.

### Example

```python
style='time'
```

### Output

* Lunch → Circle marker
* Dinner → Triangle marker

### Use Case

Useful for distinguishing groups using different marker styles.

---

## 4. size

### Purpose

Changes the size of points according to a numerical variable.

### Example

```python
size='size'
```

### Output

* Small value → Small point
* Large value → Large point

### Use Case

Adds an additional numerical dimension to the graph.

---

# Other Important Parameters

## 5. col

### Purpose

Creates separate plots column-wise.

### Example

```python

```

### Output

Creates separate graphs for each category.

---

## 6. row

### Purpose

Creates separate plots row-wise.

### Example

```python

```

### Output

Creates separate rows of graphs for each category.

---

## 7. col_wrap

### Purpose

Controls how many plots appear in one row.

### Example

```python

```

### Use Case

Improves layout when many subplots are present.

---

## 8. palette

### Purpose

Changes the color theme of the graph.

### Example

```python

```

### Popular Palettes

* deep
* viridis
* Set1
* coolwarm
* pastel

## 9. alpha

### Purpose

Controls transparency of points.

### Example

```python

```

### Range

* 0 → Fully transparent

---

## 10. height

### Purpose

Controls the height of each subplot.

### Example

```python

```

---

## 11. aspect

### Purpose

Controls width-to-height ratio.

### Example

```python

```

## c. Boxplot

# Box Plot

**Definition:**
A Box Plot is used to visualize the distribution, spread, and outliers of numerical data.

## Components

```text
Min    Q1    Median    Q3    Max
 |------|====|====|------|
        < Box >
```

* **Min** → Minimum value
* **Q1** → 25% of data
* **Median** → Middle value (50%)
* **Q3** → 75% of data
* **Max** → Maximum value

### Outliers

```text
•      |------|====|====|------|      •
Outlier                              Outlier
```

Outliers are unusual values that lie far from the rest of the data.

## Uses

* Detect outliers
* Check data spread
* Compare distributions

## One-Line Note

**Box Plot → Shows the distribution of numerical data using Min, Q1, Median, Q3, Max, and Outliers.**

# PairGrid

````python
g = sns.PairGrid(data=iris, hue='species')
g.map(sns.scatterplot)
````

````python
# map_diag - take diagonal data
# map_offdiag - no take digonal data

g = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
````

````python

g = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
plt.show()
````
